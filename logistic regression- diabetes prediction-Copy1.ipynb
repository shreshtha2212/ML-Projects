## Outlier detection

```python
X=df.drop(columns='Outcome',axis=1)
y=df['Outcome']
```

## Outlier detection -> Box plot

```python
fig, ax=plt.subplots(figsize=(15,15))
sns.boxplot(data=X, ax=ax)
plt.savefig('boxplot.jpg')
```

```python
X.shape
```

```python
y.shape
```

```python
cols = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 
        'BMI', 'DiabetesPedigreeFunction', 'Age']

outlier_flags = pd.DataFrame(False, index=X.index, columns=cols)

# Flag outliers column-by-column
for col in cols:
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outlier_flags[col] = (X[col] < lower_bound) | (X[col] > upper_bound)

# Count how many columns are outliers per row
outlier_count = outlier_flags.sum(axis=1)

# Remove rows that are outliers in 2 or more columns
mask = outlier_count < 2

X_outlier_detection = X[mask]
y_outlier_detection = y[mask]


```

## Model Training

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test= train_test_split(X_scaled, y_outlier_detection, test_size=0.33, random_state=42)
```

```python
y_train.value_counts()
```

```python
# we see above the data is biased as 0 is more than 1 so the tricks to reduce this is 1) oversampling take the minority class and increase that number to the majority class 2) undersampling we will take the majority class and reduce that number till the minority class number
```

## SMOTE technique

```python
from imblearn.over_sampling import SMOTE
smote=SMOTE(random_state=42)
X_train_resampled, y_train_resampled= smote.fit_resample(X_train, y_train)

print("\nResampled class distribution:")
print(pd.Series(y_train_resampled).value_counts())
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier()
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
knn=KNeighborsClassifier()
```

```python
knn.fit(X_train_resampled, y_train_resampled)
```

## Model Predictions

```python
y_predictions_knn=knn.predict(X_test)
y_predictions_knn
```

```python
print("Confusion Matrix")
print(confusion_matrix(y_test,y_predictions_knn))
```

```python
print("Classification Report")
print(classification_report(y_test,y_predictions_knn))
```
