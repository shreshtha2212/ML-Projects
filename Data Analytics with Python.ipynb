## Outlier

```python
# methods to treat an outlier:1)trimming 2)capping
```

```python
# ways of finding an outlier: 1) using z score  2) using interquartile range  3)using percentile
```

```python
# if we have normal distribution then we will use z score and if we have skewed distribution then we will use interquartile range and if we have any other distribution other than skewed or normal then we will use percentile 
```

```python
# normal distribution: it is bell shaped curve where mean is the center of that curve and most of the data is clustered around the center and as we move far away from the center then we will get less data  68% of the data lies between mu-sigma and mu+sigma interval between mu-2sigma and mu+2sigma i will get 95% of my data between mu+3sigma and mu-3sigma i will get 99.7% of my data if i move away from3 standard deviation then i will get outliers 
```

```python
import numpy as np
import seaborn as sns
```

```python
from matplotlib import pyplot as plt
```

```python
data=pd.read_csv("Scholarship.csv")
```

```python
df=data.copy()
```

```python
df.head()
```

```python
df.shape
```

```python
# now to check for the outliers we have tosee whcih of the columns follows a normal dist 
```

```python
plt.subplot(1,2,1)
sns.distplot(df["semester_percentage"])
plt.subplot(1,2,2)
sns.distplot(df["scholarship_exam_marks"])
plt.show()
```

```python
#semester_percentage is a normal dist while scholarship_exam_marks is a skewed dist
```

```python
mean=df['semester_percentage'].mean()
std=df['semester_percentage'].std()
```

```python
print(mean,',',std)
```

```python
highest_allowed=mean+3*std
lowest_allowed=mean-3*std
```

```python
print("Highest allowed:",highest_allowed," and lowest allowed:",lowest_allowed)
```

## Trimming

```python
df[(df["semester_percentage"]<lowest_allowed) | (df["semester_percentage"]>highest_allowed)]
```

```python
df['z-score']=(df["semester_percentage"]-df["semester_percentage"].mean())/df["semester_percentage"].std() # (x-meu)/std dev
```

```python
df.head()
```

```python
df[(df['z-score']<-3) & (df['z-score']>3)]
```

```python
df[(df['z-score']>-3) & (df['z-score']<3)]
```

## Capping

```python
upper_limit=mean+3*std
lower_limit=mean-3*std
```

```python
df["semester_percentage"]=np.where(df["semester_percentage"]>upper_limit,upper_limit,np.where(df["semester_percentage"]<lower_limit, lower_limit,df["semester_percentage"]) ) #if condition satisfies we will go withx else y
```
