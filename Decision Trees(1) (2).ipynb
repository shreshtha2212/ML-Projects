```python
df['fixed acidity']=df['fixed acidity'].replace(0,df['fixed acidity'].median())
df['volatile acidity']=df['volatile acidity'].replace(0,df['volatile acidity'].median())
df['citric acid']=df['citric acid'].replace(0,df['citric acid'].median())
df['residual sugar']=df['residual sugar'].replace(0,df['residual sugar'].median())
df['chlorides']=df['chlorides'].replace(0,df['chlorides'].median())
df['free sulfur dioxide']=df['free sulfur dioxide'].replace(0,df['free sulfur dioxide'].median())
df['total sulfur dioxide']=df['total sulfur dioxide'].replace(0,df['total sulfur dioxide'].median())


df['density']=df['density'].replace(0,df['density'].mean())

df['pH']=df['pH'].replace(0,df['pH'].mean())

df['sulphates']=df['sulphates'].replace(0,df['sulphates'].median())

df['alcohol']=df['alcohol'].replace(0,df['alcohol'].median())
```

## Outlier detection

```python
fig, ax=plt.subplots(figsize=(15,15))
sns.boxplot(data=X, ax=ax)
plt.savefig('boxplot.jpg')
cols = ['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar', 'chlorides', 'free sulfur dioxide', 'total sulfur dioxide', 'density', 'pH', 'sulphates', 'alcohol'  ]

outlier_flags = pd.DataFrame(False, index=X.index, columns=cols)

# Flag outliers column-by-column
for col in cols:
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outlier_flags[col] = (X[col] < lower_bound) | (X[col] > upper_bound)

# Count how many columns are outliers per row
outlier_count = outlier_flags.sum(axis=1)

# Remove rows that are outliers in 2 or more columns
mask = outlier_count < 2

X_outlier_detection = X[mask]
y_outlier_detection = y[mask]
```

## Standardization

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
X_scaled=scaler.fit_transform(X_outlier_detection)
X_scaled=pd.DataFrame(X_scaled,columns=cols)
X_scaled.describe()
```

## split the data into training and testing set

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test= train_test_split(X_scaled, y_outlier_detection, test_size=0.33, random_state=42)
```

```python
y_train.value_counts()
```

```python
from imblearn.over_sampling import RandomOverSampler

ros = RandomOverSampler(random_state=42)
X_train_resampled, y_train_resampled = ros.fit_resample(X_train, y_train)

print("\nResampled class distribution:")
print(pd.Series(y_train_resampled).value_counts())
```

```python
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier()
model.fit(X_train_resampled, y_train_resampled)
```

## Model predictions

```python
y_pred=model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn import tree
plt.figure(figsize=(20,20))
tree.plot_tree(model, filled=True)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_pred,y_test))
```

## Random forest classifier

```python
rf=RandomForestClassifier()
rf.fit(X_train_resampled,y_train_resampled)
```

```python
rf.score(X_test,y_test)
```
