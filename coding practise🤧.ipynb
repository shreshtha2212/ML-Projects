**Number of digits in factorial**

```python
import math

def num_digits_in_factorial(n):
    if n == 0 or n == 1:
        return 0 # 0! and 1! both have 1 digit
    else:
        return math.log10(n) + num_digits_in_factorial(n - 1)

def count_digits(n):
    return math.floor(num_digits_in_factorial(n)) + 1
```

**Exactly 3 divisors**

```python
def isPrime(N):
    if(N==1):
        return False
    for i in range(2,1+int(math.sqrt(N))):
        if N%i==0:
            return False
    return True
def exactly3Divisors(N):
    N=int(math.sqrt(N))
    c=0
    for i in range(1,N+1):
        if(isPrime(i)):
            c=c+1
    return c
```

**Quadratic equation roots**

```python
def quadracticRoots(a,b,c):
    roots=[]
    root1=0
    root2=0
    temp=pow(b,2)-(4*a*c)
    if temp<0:
        roots.append(-1)
    else:
        root1=math.floor((-b+math.sqrt(temp))/(2*a))
        root2=math.floor((-b-math.sqrt(temp))/(2*a))
        roots.append(max(root1,root2))
        roots.append(min(root1,root2))
    return roots
```

**count set bits in first N natural numbers**

```python
#suppose N is 11 then we have 0-0000, 1-0001, 2-0010, 3-0011, 4-0100, 5-0101, 6-0110, 7-0111, 8-1000, 9-1001, 10-1010, 11-1011
# we see that the largest power of 2 which is less than equal to 11 is 3 so the number of set bits before 2 power 3 that is 8 will be
# in the 0th bit we have 8/2 bits set, in the 1st bit we have 8/2 bits set and in 2nd bit we again have 8/2 bits set
# so we can say that we have 3*4=12 bits set before 2 power 3 which we can also represent as 2 power (3 minus 1) * 3
# now for the next part we are left with 4 combinations 1000,1001,1010,1011 in these combinations we will separate the msb that is 1
# from the rest of the portion of the combinations that is we will be left with 000,001,010,011
# to account for the 4 msbs we will represent as 11-(2 power 3)+1 which will give us 4 
# to account for the rest of the portion we will represent as 11-(2 power 3) and we will pass the rest of bits through the function again recursively
#we will create another function to compute the power of 2 less than or equal to N
```

```python
def pow2(N):
    x=0
    while((1<<x)<=N):
        x+=1
    return x-1
        
def solve(N):
    if N==0:
        return 0
    x=pow2(N)
    f=x*(1<<(x-1))
    g=N-(1<<x)+1
    h=N-(1<<x)
    ans=f+g+solve(h)
    return ans
```

**rope cutting problem**

```python
def rope(n,a,b,c):
    if(n<0):
        return -1
    if(n==0):
        return 0
    res=max(rope(n-a,a,b,c), rope(n-b,a,b,c), rope(n-c, a,b,c))
    if(res==-1):
        return -1
    return res+1
```

**substring of string**

```python
def swap(s, a, b):
    s = list(s)  # Convert string to a list
    s[a], s[b] = s[b], s[a]  # Swap characters
    return "".join(s)  # Convert back to string

def permute(s, l, r):
    if l > r:
        print(s)  # Print permutation
        return
    
    for i in range(l, r + 1):
        s = swap(s, l, i)  # Swap current element with `l`
        permute(s, l + 1, r)  # Recursive call
        s = swap(s, l, i)  # Backtrack to restore original state

# Example usage:
s = "ABC"  # Input string
permute(s, 0, len(s) - 1)
```

**josephus problem**

```python
def fun(n, k):
    if n == 1:
        return 0  # Base case: The survivor's index in a 1-person circle is always 0
    x = fun(n - 1, k)  # Recursive call
    y = (x + k) % n  # Josephus position formula
    return y 
```

**subset sum**

```python
def sub(arr,n,s):
    if(s==0):      # writing base condn for s before base condn for n is key here otherwise our answer will be 1 instead of 2 for the case sub([1,2,3],3,3) 
        return 1
    if(n==0):
        return 0
    
    if(arr[n-1]<=s):
        return sub(arr,n-1,s-arr[n-1])+sub(arr,n-1,s)
    else:
        return sub(arr,n-1,s)
```

**trapping rainwater**

```python
def trap(arr):
    l = 0
    r = len(arr) - 1
    left = arr[l]
    right = arr[r]
    ans = 0

    while l < r:
        if left < right:
            l += 1
            if arr[l] < left:
                ans += left - arr[l]
            else:
                left = arr[l]
        else:
            r -= 1
            if arr[r] < right:
                ans += right - arr[r]
            else:
                right = arr[r]

    return ans
```

```python
trap([0,1,0,2,1,0,1,3,2,1,2,1])
```

# maximum circular subarray sum

```python
def ar(arr):
    curMaxi=0
    curMini=0
    globMax=arr[0]
    globMin=arr[0]
    for i in range(0,len(arr)):
        curMaxi=max(arr[i],arr[i]+curMaxi)
        curMini=min(arr[i],arr[i]+curMini)
        globMax=max(globMax,curMaxi)
        globMin=min(globMin,curMini)
    if globMax>0:
        return max(globMax, (sum(arr)-globMin)) 
```

# Sliding window problem

```python
def slide(arr,k):
    cur=sum(arr[0:k])
    s=cur
    for i in range(k,len(arr)):
        cur=cur+arr[i]-arr[i-k]
        if(cur>s):
            s=cur
    return s
```

```python
slide([1,2,3,4,5],3)
```

# Binary search

```python
def bSearch(arr,x,low,high):
    if(low>high):
        return -1
    mid=(low+high)//2
    if(arr[mid]==x):
        return mid
    elif arr[mid]>x:
        low=mid+1
        bSearch(arr,x,mid+1,high)
    else:
        high=mid-1
        bSearch(arr,x,low,high)
    return -1
```

```python
bSearch([2,3,4,5],3,0,3)
```

# Index of first occurence in sorted array

```python
def firstocc(arr,n,x):
    low=0
    high=n-1
    while(low<=high):
        mid=(low+high)//2
        if(x>arr[mid]):
            low=mid+1
        elif x<arr[mid]:
            high=mid-1
        else:
            if mid==0 or arr[mid-1]!=arr[mid]:
                return mid
            else:
                high=mid-1
    return -1
        
```

# lomuto

```python
def lomuto(arr,l,h):
    pivot=arr[h]
    i=l-1
    for j in range(l,h):
        if(arr[j]<pivot):
            i+=1
        arr[i],arr[j]=arr[j],arr[i]
    arr[i+1],arr[h]=arr[h],arr[i+1]
    return i+1
```

```python
def maxi(n, arr, i):
    # Reheapify the heap to maintain the max-heap property
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        maxi(n, arr, largest)

def buildHeap(arr):
    n = len(arr)
    start_idx = (n // 2) - 1  # Start from the last non-leaf node
    
    for i in range(start_idx, -1, -1):
        maxi(n, arr, i)

def heapSort(arr):
    n = len(arr)
    
    # Step 1: Build a max-heap
    buildHeap(arr)

    # Step 2: Extract elements one by one and move them to the end of the array
    for i in range(n-1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]  # Swap the root (largest) with the last element
        maxi(i, arr, 0)  # Re-heapify the reduced heap (excluding the sorted part)
    
    return arr

# Example usage:
arr = [50, 20, 10, 4, 15]
sorted_arr = heapSort(arr)
print(sorted_arr)
```

# longest subarray with given sum

```python
def slide(arr, su):
    r = {0: -1}  # Store prefix sum as keys and indices as values
    res = 0
    s = 0
    
    for i in range(len(arr)):
        s += arr[i]
        
        if s - su in r:  # Check if (current prefix sum - target) exists
            res = max(res, i - r[s - su])
        
        if s not in r:  # Store only the first occurrence of a prefix sum
            r[s] = i
    
    return res
```

```python
ar=[1,9,3,4,2,20]

def sub(ar):
    ar=set(ar)
    ar=list(ar)
    maxi=0
    res=0
    for i in range(len(ar)):
        if(i-1 in ar):
            i+=1
        else:
            j=ar[i]
            res+=1
            j+=1
            while(j in ar):
                j=j+1
                res+=1
            if(res>maxi):
                maxi=res
            res=0
    return maxi
```

# check if a string is subsequence of other

```python
def isSub(s1, s2, m, n):
    if n == 0:  # If s2 is empty, it's always a subsequence
        return True
    if m == 0:  # If s1 is empty but s2 is not, it's not a subsequence
        return False
    if s1[m - 1] == s2[n - 1]:  # If last characters match, move both back
        return isSub(s1, s2, m - 1, n - 1)
    else:  # Otherwise, move only s1 back
        return isSub(s1, s2, m - 1, n)

# Example usage:
s1 = "abcdef"
s2 = "ace"
print(isSub(s1, s2, len(s1), len(s2)))  # True
```

```python
def naivePat(s,p):
    n=len(s)
    m=len(p)
    for i in range(n-m+1):
        j=0
        while(j<m):
            
            if(s[i+j]!=p[j]):
                break
            j+=1
        if(j==m):
            print(i)
```

```python
def fill(arr):
    n = len(arr)
    p = [0] * n  # Initialize p with size n
    l = 0
    i = 1
    while i < n:
        if arr[i] == arr[l]:
            l += 1
            p[i] = l
            i += 1
        else:
            if l > 0:
                l = p[l-1]  # Fallback to the previous longest prefix
            else:
                p[i] = 0
                i += 1
    return p
```

# insert at given position

```python
def pos(head, p, a):
    j = Node(a)
    
    # Inserting at head (position 0)
    if p == 0:
        j.next = head
        return j
    
    curr = head
    k = 0
    
    # Traverse to the (p-1)th node
    while k < p - 1 and curr is not None:
        curr = curr.next
        k += 1
    
    # Check if position is valid
    if curr is None:
        print("Position out of bounds.")
        return head
    
    # Standard insertion
    j.next = curr.next
    curr.next = j
    
    return head
```

# sorted insert linked list

```python
def ins(head, a):
    p = Node(a)

    # Case 1: Insert before head
    if head is None or a < head.key:
        p.next = head
        return p

    # Case 2: Traverse to find insertion point
    curr = head
    while curr.next is not None and curr.next.key < a:
        curr = curr.next

    # Insert p after curr
    p.next = curr.next
    curr.next = p

    return head
```

```python
def rev(head):
    curr = head
    prev = None
    while curr is not None:
        curr.next, curr.prev = curr.prev, curr.next
        prev = curr            # track the last node processed
        curr = curr.prev       # move to the next (original prev)
    return prev                # new head of reversed list
```

```python
class deque:
    def __init__(self,c):
        self.l=[None]*c
        self.cap=c
        self.sz=0
        self.front=0
    def insertRear(self,x):
        if self.sz==self.cap:
            return None
        new_rear=(self.front+self.sz)%self.cap
        self.l[new_rear]=x
        self.sz+=1
    def deleteRear(self):
        sz=self.sz
        if sz==0:
            return
        rear=(self.front+sz-1)%self.cap
        self.sz-=1
        return self.l[rear]
    def insertFront(self,x):
        if self.sz==self.cap:
            return
        else:
            self.front=(self.front-1)%self.cap # when inserting in front initially when front=0 (-1%5) will give us 5-1=4 so the new element will be inserted at the end of the list as modulo in python always gives positive result
            self.l[self.front]=x
            self.sz+=1
            
    def deleteFront(self):
        if self.sz==0:
            return None
        res=self.l[self.front]
        self.front=(self.front+1)%self.cap
        self.sz-=1
        return res


    
```

# detect cycle in undirected graph

```python
from collections import deque
def cycle(adj,s):
    visited=[False]*len(adj)
    d=[0]*len(adj)
    q=deque()
    q.append([s,-1])
    while(q):
        u=q.popleft()
        if visited[u[0]-1]:
            continue
        visited[u[0]-1]=True
        
        for i in adj[u[0]-1]:
            j=i-1
            
            d[j]+=1
            q.append([j,u[0]])
        
    return d
```

# longest common subsequence using top down

```python
def lcs(s1,s2,l1,l2,visited):
    if visited[l1][l2]!=-1:
        return visited[l1][l2]
    if l1==0 or l2==0:
        return 0
    if s1[l1-1]==s2[l2-1]:
        visited[l1][l2]=1+lcs(s1,s2,l1-1,l2-1,visited)
    if s1[l1-1]!=s2[l2-1]:
        visited[l1][l2]=max(lcs(s1,s2,l1,l2-1,visited),lcs(s1,s2,l1-1,l2,visited))
        
    return visited[l1][l2]
```

```python
def f(s1,s2,n1,n2):
    if n2<0 and n1>0:
        return abs(len(s1)-len(s2))-1
    if n2<0 and n1<0:
        return 0 
    if s2[n2]!=s1[n1]:
        if s2[n2] in s1[n1:]:
            s1[n1]=''
            return 1+f(s1,s2,n1-1,n2-1)
        else:
            s1.replace(s1[n1],s2[n2])
            return 1+f(s1,s2,n1-1,n2-1)
    if s2[n2]==s1[n1]:
        return f(s1,s2,n1-1,n2-1)
```

```python
def edit_distance(s1, s2):

    n1, n2 = len(s1), len(s2)

    dp = [[0 for _ in range(n2+1)] for _ in range(n1+1)]

 

    for i in range(n1+1):

        for j in range(n2+1):

            if i == 0:

                dp[i][j] = j  # Insert all s2 characters

            elif j == 0:

                dp[i][j] = i  # Remove all s1 characters

            elif s1[i-1] == s2[j-1]:

                dp[i][j] = dp[i-1][j-1]  # Characters match

            else:

                dp[i][j] = 1 + min(

                    dp[i-1][j],    # Deletion

                    dp[i][j-1],    # Insertion

                    dp[i-1][j-1]   # Substitution

                )

    return dp[n1][n2]
```

# longest increasing subsequence

```python
def lis(i, prev_index, a,dp):
    if i == len(a):
        return 0
    if dp[i][prev_index+1]!=-1:
        return dp[i][prev_index+1]

    # Option 1: skip the current element
    not_take = lis(i + 1, prev_index, a,dp)

    # Option 2: take it if it's greater than the previous element
    take = 0
    if prev_index == -1 or a[i] > a[prev_index]:
        take = 1 + lis(i + 1, i, a,dp)

    dp[i][prev_index+1]=max(take, not_take)
    return dp[i][prev_index+1]



    
```

```python
#Back-end complete function Template for Python 3
import math
class Solution:
    # function to check if N is prime
    def isPrime(self,N):
        if (N==1):
            return False
        for i in range(2,1+int(math.sqrt(N))):
            if N%i==0:
                return False
        return True
        
    # function to check the numbers with exactly 3 divisors
    def exactly3Divisors(self,N):
        N = int(math.sqrt(N))
        #Initializing counter to zero
        counter=0 
        #Running a loop from 1 to sqrt(N)
        for i in range(1,N+1): 
        
            # A number N only has 3 divisors if it is a  
            # perfect square and its square  root is a prime number, 
            # and we know the number of perfect squares less than N which 
            # is sqrt(N), so just checking if i is prime or not
            if(self.isPrime(i)): 
                counter+=1
        return counter
```
